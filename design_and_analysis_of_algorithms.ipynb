## **Find the aximum value in an array**

```python
def find_max(a):


    current_max = a[0]
     # Start with the first element

    for num in a:
        if num > current_max:
            current_max = num  # Found a bigger one!

    return current_max

#print the max
numbers = [2, 5, 3, 6, 7, 4, 9, 8]
print(find_max(numbers))  # Output: 9
```

## **Find the minimum value in an array**

```python
def find_min(b):
  current_min = b[0]

  for num in b:
    if num < current_min:
      current_min = num

  return current_min

#print the min
nums = [2, 5, 3, 6, 7, 4, 9, 8]

print(find_min(nums))
```
